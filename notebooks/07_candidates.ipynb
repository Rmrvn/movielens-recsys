```python
%load_ext autoreload
%autoreload 2

import sys
import time
sys.path.insert(0, '../src')

from recsys.data import load_ratings, make_positive, temporal_split, user_items
from recsys.metrics import evaluate
from recsys.ranking import  build_candidates

ratings = load_ratings('../data')
pos = make_positive(ratings)
train, test = temporal_split(pos)
train_core, valid = temporal_split(train)

params = {'factors': 64, 'reg': 100, 'alpha': 10}
valid_truth = user_items(valid)
users = list(valid_truth)

start = time.time()
cand = build_candidates(train_core, users, params, n_candidates=100, target=valid)
print(f'{time.time() - start:.1f} сек')
print('строк:', len(cand), '| доля положительных:', round(cand['label'].mean(), 4))
print(cand.head())

```
